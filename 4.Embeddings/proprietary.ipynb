```python
import os
from dotenv import load_dotenv
from langchain_openai import OpenAIEmbeddings
from langchain_community.document_loaders import PyPDFLoader
from pypdf import PdfReader
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
load_dotenv()
```

```python
embeddings = OpenAIEmbeddings(
    model="text-embedding-3-small"
)
embeddings2 = OpenAIEmbeddings(
    model="text-embedding-3-large"
)
```

```python
text = "I forgot my login password."

vector = embeddings.embed_query(text)

print("Output type:", type(vector))
print("Dimensions:", len(vector))
print("First 5 values:", vector[:5])
```

```python
file_path = './Document Loaders in LangChain-10.pdf'
reader = PdfReader(file_path)
docs=[]
for page_number, page in enumerate(reader.pages, start=1):
    text = page.extract_text() or ""

    if text.strip():
        doc = Document(
            page_content=text,
            metadata={
                "source": file_path,
                "page": page_number,
            },
        )
        docs.append(doc)

print("Total PDF pages:", len(reader.pages))
print("Documents created:", len(docs))
```

```python
if docs:
    print("Metadata:", docs[0].metadata)
    print("Text:\n", docs[0].page_content[:1000])
else:
    print("No text extracted. Scanned PDFs may need OCR.")
```

```python
# 2. Create chunks
splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=150,
)

chunks = splitter.split_documents(docs)

print("Total chunks:", len(chunks))
```

```python
# 3. Generate embeddings using your OpenAI embeddings object
if not chunks:
    raise ValueError("No text extracted. Check the PDF; scanned pages may need OCR.")

texts = [chunk.page_content for chunk in chunks]

vectors = embeddings.embed_documents(texts)

print("Total vectors:", len(vectors))
print("Dimensions per vector:", len(vectors[0]))
print("First vector preview:", vectors[0][:5])
print("First chunk metadata:", chunks[0].metadata)
```
