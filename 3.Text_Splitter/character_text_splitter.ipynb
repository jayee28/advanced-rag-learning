```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text = """
Retrieval-Augmented Generation, commonly known as RAG, is a technique that combines information retrieval with large language models. Instead of depending only on the knowledge stored in the model's parameters, a RAG system searches an external knowledge base and retrieves information relevant to the user's query.
The retrieved information is then added to the prompt as context. The language model uses the query and retrieved context to generate a more accurate and grounded response. This approach can reduce hallucination and allows applications to answer questions using private, domain-specific, or recently updated information.
A typical RAG pipeline contains document loaders, text splitters, embedding models, vector stores, retrievers, prompts, and language models. Document loaders collect data from sources such as PDFs, websites, JSON files, and databases. Text splitters divide large documents into smaller chunks so they can be embedded, stored, and retrieved efficiently.
"""
```

```python
splitter = CharacterTextSplitter(
    separator='',
    chunk_size=100,
    chunk_overlap=20,
    length_function=len
)
```

```python
from termcolor import COLORS, colored
from random import choice
```

```python
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Numer of chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks,1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
def create_chunks(text:str, chunk_size:int, separator:str, chunk_overlap:int=0)->list[str]:
    splitter = CharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separator=separator,
    )
    chunks = splitter.split_text(text=text)
    return chunks
```

```python
display_chunks(create_chunks(text, 100, '',20))
```

```python
# chunking based on paragraph
display_chunks(create_chunks(text, 100, '\n\n', 20))
```

```python
def display_chunks_with_overlaps(chunks, chunk_overlap):
    if not chunk_overlap:
        raise(ValueError("Chunk Overlap cannot be 0"))
    
    # create vars
    overlap_text = ""
    overlap_color = "white"
    colors_list = list(COLORS.keys())[2:8]
    
    print(f"Total Number of Chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        if num == 1:
            # print the chunk without overlap
            non_overlap_text = chunk[0:-chunk_overlap]
            overlap_text = chunk[-chunk_overlap:]
            print(colored(non_overlap_text, choice(colors_list)),
                  colored(overlap_text,overlap_color),
                  sep="",end="\n\n")
        else:
            # print the overlap text at start of chunk
            print(colored(overlap_text, overlap_color), end="")
            # print non overlap text
            non_overlap_text = chunk[chunk_overlap:-chunk_overlap]
            print(colored(non_overlap_text, choice(colors_list)), end="")
            # print overlap text at end of chunk
            overlap_text = chunk[-chunk_overlap:]
            print(colored(overlap_text, overlap_color), end="\n\n")
```

```python
display_chunks(chunks=create_chunks(text=text, chunk_size=100, separator=''))
```

```python
display_chunks_with_overlaps(chunks=create_chunks(text=text, chunk_size=100, separator='',chunk_overlap=10),chunk_overlap=10)
```
