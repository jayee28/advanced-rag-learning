```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text = """
Retrieval-Augmented Generation, commonly known as RAG, is a technique that combines information retrieval with large language models. Instead of depending only on the knowledge stored in the model's parameters, a RAG system searches an external knowledge base and retrieves information relevant to the user's query.
The retrieved information is then added to the prompt as context. The language model uses the query and retrieved context to generate a more accurate and grounded response. This approach can reduce hallucination and allows applications to answer questions using private, domain-specific, or recently updated information.
A typical RAG pipeline contains document loaders, text splitters, embedding models, vector stores, retrievers, prompts, and language models. Document loaders collect data from sources such as PDFs, websites, JSON files, and databases. Text splitters divide large documents into smaller chunks so they can be embedded, stored, and retrieved efficiently.
"""
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10,
    separators=["\n\n", "\n", " ", ""]
)
```

```python
splitter.split_text(text)
```

```python
example_text = """Artificial intelligence is transforming technology and shaping the future.

Machine learning algorithms are becoming more sophisticated every day. 
Deep learning models can now process vast amounts of data efficiently.

Natural language processing has made significant strides in recent years.
Transformers architecture revolutionized the field in 2017.
Models like GPT and BERT have set new benchmarks.

Computer vision systems can now identify objects with remarkable accuracy.
Convolutional neural networks excel at image recognition tasks.
Self-driving cars rely heavily on advanced computer vision.

The impact of AI extends across multiple industries including healthcare, finance, and transportation.
Ethical considerations around AI development are becoming increasingly important.
Researchers are working on making AI systems more transparent and explainable."""
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=150,
    chunk_overlap=20,
    separators=["\n\n", "\n", " ", ""]
)
```

```python
chunks = splitter.split_text(example_text)
print(chunks)
```

```python
from termcolor import COLORS, colored
from random import choice
```

```python
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]
    print(f"Total Number of Chunks: {len(chunks)}")
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
display_chunks(chunks)
```

```python
from langchain_core.documents import Document
```

```python
# create the list of documents

list_of_text = [text, example_text]

docs = [Document(page_content=text) for text in list_of_text]

print(docs)
```

```python
# split the documents

chunks_docs = splitter.split_documents(docs)

print(chunks_docs)
print(len(chunks_docs))
```
