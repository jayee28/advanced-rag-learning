```python
from langchain_community.document_loaders import JSONLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path('./knowledge-source/apparels.json')
file_path.exists()
```

```python
def metadata_func(record, metadata):
    metadata["product_id"] = record.get("productID")
    metadata["product_name"] = record.get("productName")
    metadata["manufacturer"] = record.get("manufacturer")
    metadata["price"] = record.get("price")
    metadata["category"] = record.get("category")
    metadata["url"] = record.get("Url")

    del metadata['seq_num']

    return metadata
```

```python
loader = JSONLoader(file_path=file_path, jq_schema='.products[]', content_key="Description", metadata_func=metadata_func)
```

```python
documents = loader.load()
```
