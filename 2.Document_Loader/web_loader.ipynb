```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader(
    "https://www.example.com"
)

documents = loader.load()
```

```python
print("Total documents:", len(documents))
print(type(documents))
print(type(documents[0]))
```

```python
for i, doc in enumerate(documents, start=1):
    print(f"--- Document {i} ---")
    print("Content:")
    print(doc.page_content[:1000])

    print("\nMetadata:")
    print(doc.metadata)
```

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
# create the loader
base_url =  "https://docs.langchain.com/oss/python/integrations/document_loaders"
loader = RecursiveUrlLoader(url=base_url, max_depth=2)
documents = loader.load()
```

```python
len(documents)
```

```python
for i in range(10):
    doc = documents[i]
    print(doc.metadata, end="\n\n")
```

```python
# lazy load
documents_lazy_load = loader.lazy_load()
```

```python
documents_lazy_load
```

```python
counter = 0
for documents in documents_lazy_load:
    if counter ==20:
        break
    counter += 1
    print(documents.page_content[:3000])
    print(documents.metadata)
```
